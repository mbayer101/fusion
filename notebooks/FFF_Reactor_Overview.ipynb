```python
# Inspect live 3D OpenFOAM Turbulent CHT Temperature Fields
def read_foam_scalar(path):
    with open(path) as f:
        txt = f.read()
    m = re.search(r'internalField\s+nonuniform\s+List<scalar>\s+\d+\s*\((.*?)\)', txt, re.S)
    return np.fromstring(m.group(1), sep='\n') - 273.15

if os.path.exists('../openfoam_cht/60/fluid_lbe/T'):
    T_lbe = read_foam_scalar('../openfoam_cht/60/fluid_lbe/T')
    T_fw  = read_foam_scalar('../openfoam_cht/60/solid_fw/T')
    print(f'3D Fluid LBE Temp (Min/Mean/Max): {T_lbe.min():.2f} C / {T_lbe.mean():.2f} C / {T_lbe.max():.2f} C')
    print(f'3D Solid FW  Temp (Min/Mean/Max): {T_fw.min():.2f} C / {T_fw.mean():.2f} C / {T_fw.max():.2f} C')
```
